# Task 2: Document Processing and Text Chunking

```python
from langchain_community.document_loaders import TextLoader
import glob

document_files = glob.glob("/content/rag_documents/documents/*.txt")

documents = []

for file_path in document_files:
    loader = TextLoader(file_path, encoding="utf-8")
    documents.extend(loader.load())

print("Number of documents loaded:", len(documents))
```

```python
for i, doc in enumerate(documents):
    print(f"\nDocument {i + 1}")
    print("Source:", doc.metadata.get("source"))
    print("Characters:", len(doc.page_content))
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks = text_splitter.split_documents(documents)

print("Total chunks created:", len(chunks))
```

```python
for i, chunk in enumerate(chunks[:5]):
    print(f"\n--- Chunk {i + 1} ---")
    print(chunk.page_content)
```

# Task 3: Create Document Embeddings

```python
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

print("Embedding model loaded successfully!")
```

# Task 4: Vector Database Creation

```python
from langchain_community.vectorstores import FAISS

vectorstore = FAISS.from_documents(
    chunks,
    embeddings
)

print("FAISS Vector Database Created Successfully!")
```

```python
vectorstore.save_local("/content/faiss_index")

print("FAISS Vector Database Saved Successfully!")
```

```python
question = "What is Machine Learning?"

retrieved_docs = vectorstore.similarity_search(
    question,
    k=1
)

print("Question:", question)

if retrieved_docs:
    print("\nRetrieved Information:")
    print(retrieved_docs[0].page_content)
else:
    print("\nNo relevant information found.")
```

# Task 6: RAG Chatbot

```python
retriever = vectorstore.as_retriever(
    search_kwargs={"k": 1}
)

print("Retriever created successfully!")
```

```python
def rag_chatbot(question):
    retrieved_docs = retriever.invoke(question)

    if not retrieved_docs:
        return "I could not find relevant information in the knowledge base."

    context = retrieved_docs[0].page_content

    response = f"""
Question:
{question}

Answer based on the knowledge base:
{context}
"""

    return response
```

```python
question = "What is Machine Learning?"

answer = rag_chatbot(question)

print(answer)
```

```python
question = "What is Python?"

answer = rag_chatbot(question)

print(answer)
```

# Task 7: Interactive RAG Chatbot

```python
print("🤖 RAG Chatbot")
print("Ask questions about AI/ML, Python, or Deep Learning.")
print("Type 'exit' to stop.\n")

while True:
    question = input("You: ")

    if question.lower() == "exit":
        print("Chatbot: Goodbye!")
        break

    answer = rag_chatbot(question)

    print("\nChatbot:")
    print(answer)
    print()
```
